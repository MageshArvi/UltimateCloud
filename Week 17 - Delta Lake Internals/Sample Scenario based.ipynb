```python
%sql
create volume if not exists delta_lake_study_demo.delta_lake_demo.delta_sample_vol
```

```python
data = [(1,"Arvi",28),
        (2,"Pavi",35),
        (3,"Saashu",26),
        (4,"Arun",25)]

cols = ["id","name","age"]

df = spark.createDataFrame(data,cols)

df.show()
```

```python
volume_path = '/Volumes/delta_lake_study_demo/delta_lake_demo/delta_sample_vol/'
df.write.format("delta").mode("overwrite").save(volume_path)
```

```python
%sql
select * from delta.`/Volumes/delta_lake_study_demo/delta_lake_demo/delta_sample_vol`
```

```python
new_data = [(5,"Ram",28),
        (6,"Ashwin",35),
        (7,"Swetha",26),
        (8,"Shruthi",25)]

df_new = spark.createDataFrame(new_data,["id","name","age"])
```

```python
df_new.write.format("delta").mode("append").save(volume_path)
```
