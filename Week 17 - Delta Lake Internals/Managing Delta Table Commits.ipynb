```python
%sql
create or replace table orders_timetravel (
    order_id bigint,
    sku string,
    product_name string,
    product_category string,
    qty int,
    unit_price decimal(10,2)
)using DELTA
tblproperties (
    delta.autoOptimize.optimizeWrite= false,
    delta.autoOptimize.autoCompact = false
)
LOCATION 'abfss://externaldata@deltalakelearnstg.dfs.core.windows.net/orders_timetravel'
```

```python
for i in range(100):
    order_id = 1000 + i
    sku = f"SKU-BURST-{i}"
    product_name = f"BURST-{i}"
    product_category = 'Demo'
    qty = 1
    unit_price = float(i%10 + 1) * 1.0
    sql = f""" INSERT INTO orders_timetravel VALUES ({order_id},'{sku}','{product_name}','{product_category}',{qty},{unit_price})"""
    spark.sql(sql)
    
```
