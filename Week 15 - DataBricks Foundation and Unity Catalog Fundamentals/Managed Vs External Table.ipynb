```sql
create external location adls_schema_level_ext_loc
URL 'abfss://unitycatalog@databricksdemoarvi.dfs.core.windows.net/schema'
WITH (credential my_credentials)
comment 'schema level location'
```

```sql
create table my_catalog2.retail.sales(
    id int,
    name string,
    price double,
    quantity int
);

insert into my_catalog2.retail.sales values (1, 'banana', 0.99, 100), (2, 'apple', 1.99, 200);
```

```sql
create external location adls_external_table_loc_ext
URL 'abfss://unitycatalog@databricksdemoarvi.dfs.core.windows.net/external'
with (credential my_credentials)
comment 'external location for external tables'
```

```sql
create table my_catalog2.retail.sales_ext(
    id int,
    name string,
    price double,
    quantity int
)
location 'abfss://unitycatalog@databricksdemoarvi.dfs.core.windows.net/external';

insert into my_catalog2.retail.sales_ext values (1, 'banana', 0.99, 100), (2, 'apple', 1.99, 200);
```

```sql
select * from my_catalog2.retail.sales_ext
```
