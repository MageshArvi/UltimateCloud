When we have large number of distinct values, then you can go with Bucketing.

It helps in 2 purposes

1. Could help in skipping data
2. Joins



```python
customer_df = spark.read.option('header', False).option('inferSchema', True) \
    .csv('abfss://unitycatalog@databricksstudysa.dfs.core.windows.net/catalog/part-00000') \
    .toDF('customer_id', 'customer_fname', 'customer_lname', 'customer_email', 'customer_password', 'customer_street', 'customer_city', 'customer_state', 'customer_zip')
```

```python
customer_df.show()
```

```python
customer_df.write.bucketBy(4, "customer_id").saveAsTable("customer")
```
