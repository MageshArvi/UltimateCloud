```python
dbutils.fs.mkdirs("/Volumes/delta_lake_study_demo/delta_lake_demo/delta_vol1/orders_vol1")
```

```python
from pyspark.sql import Row

# Create a sample DataFrame with 5 records
data = [
    Row(id=1, name="Alice", age=30, city="New York"),
    Row(id=2, name="Bob", age=25, city="San Francisco"),
    Row(id=3, name="Charlie", age=35, city="Chicago"),
    Row(id=4, name="Diana", age=28, city="Seattle"),
    Row(id=5, name="Eve", age=32, city="Boston")
]

df = spark.createDataFrame(data)
display(df)
```

```python
volume_path = "/Volumes/delta_lake_study_demo/delta_lake_demo/delta_vol1/orders_vol1"
```

```python
df.write.format("delta").mode("overwrite").save(volume_path)
```
